### Task 1: Trends in Compensation Over Age Groups


##### 1. Line Chart of Median `ConvertedCompYearly` by Age Group


- Track how the median yearly compensation (ConvertedCompYearly) changes across different age groups.

- Use a line chart to visualize these trends.


```python
## Write your code here
def age_to_numeric(age_str):
    if pd.isnull(age_str):
        return None
    if age_str == 'Under 18 years old':
        return 16
    elif age_str == '18-24 years old':
        return 21
    elif age_str == '25-34 years old':
        return 30
    elif age_str == '35-44 years old':
        return 40
    elif age_str == '45-54 years old':
        return 50
    elif age_str == '55-64 years old':
        return 60
    elif age_str == '65 years or older':
        return 70
    else:
        return None

df['Age_numeric'] = df['Age'].apply(age_to_numeric)
median_salary_by_age = df.groupby('Age_numeric')['ConvertedCompYearly'].median().reset_index()
plt.figure(figsize=(8,5))  # Set the figure size

# Draw the line chart
plt.figure(figsize=(8,5))

# Plot the line
plt.plot(
    median_salary_by_age['Age_numeric'], 
    median_salary_by_age['ConvertedCompYearly'], 
    marker='o',  # show points
    linestyle='-',  # line style
    color='blue'
)

plt.title("Median Yearly Compensation by Age Group")  # chart title
plt.xlabel("Age")  # x-axis label
plt.ylabel("Median Compensation")  # y-axis label
plt.xticks(median_salary_by_age['Age_numeric'])  # show all age groups
plt.grid(True)  # show grid
plt.show()
```

##### 2. Line Chart of Median `ConvertedCompYearly` for Ages 25 to 45


For a closer look, plot a line chart focusing on the median compensation for respondents between ages 25 and 45.


```python
## Write your code here
# Filter the dataframe for ages between 25 and 45
age_filtered = df[(df['Age_numeric'] >= 25) & (df['Age_numeric'] <= 45)]
median_salary_filtered = age_filtered.groupby('Age_numeric')['ConvertedCompYearly'].median().reset_index()
plt.figure(figsize=(8,5))

# Plot the line for ages 25–45
plt.plot(
    median_salary_filtered['Age_numeric'], 
    median_salary_filtered['ConvertedCompYearly'], 
    marker='o', 
    linestyle='-', 
    color='green'
)

plt.title("Median Yearly Compensation for Ages 25 to 45")
plt.xlabel("Age")
plt.ylabel("Median Compensation")
plt.xticks(median_salary_filtered['Age_numeric'])
plt.grid(True)
plt.show()
```

### Task 2: Trends in Job Satisfaction by Experience Level



##### 1. Line Chart of Job Satisfaction (`JobSatPoints_6`) by Experience Level



- Use a column that approximates experience level to analyze how job satisfaction changes with experience.

- If needed, substitute an available experience-related column for `Experience`.


```python
## Write your code here
df_exp = df[['YearsCodePro', 'JobSatPoints_6']].dropna()
# Group by years of experience and calculate median job satisfaction
median_jobsat_by_exp = df_exp.groupby('YearsCodePro')['JobSatPoints_6'].median().reset_index()
plt.figure(figsize=(8,5))

plt.plot(
    median_jobsat_by_exp['YearsCodePro'], 
    median_jobsat_by_exp['JobSatPoints_6'], 
    marker='o', 
    linestyle='-', 
    color='purple'
)

plt.title("Median Job Satisfaction by Experience Level")
plt.xlabel("Years of Professional Experience")
plt.ylabel("Median Job Satisfaction")
plt.xticks(median_jobsat_by_exp['YearsCodePro'][::2], rotation=45)  # show every 2nd tick to avoid crowding
plt.grid(True)
plt.show()
```

### Task 3: Trends in Job Satisfaction and Compensation by Experience


##### 1.Line Chart of Median ConvertedCompYearly Over Experience Level

- This line chart will track how median compensation (`ConvertedCompYearly`) changes with increasing experience.

- Use a column such as `WorkExp` or another relevant experience-related column.


```python
## Write your code here
df_exp_comp = df[['YearsCodePro', 'ConvertedCompYearly']].dropna()
median_comp_by_exp = df_exp_comp.groupby('YearsCodePro')['ConvertedCompYearly'].median().reset_index()
plt.figure(figsize=(8,5))

plt.plot(
    median_comp_by_exp['YearsCodePro'], 
    median_comp_by_exp['ConvertedCompYearly'], 
    marker='o', 
    linestyle='-', 
    color='orange'
)

plt.title("Median Yearly Compensation by Experience Level")
plt.xlabel("Years of Professional Experience")
plt.ylabel("Median Yearly Compensation")
plt.xticks(median_comp_by_exp['YearsCodePro'][::2], rotation=45)  # Show every 2nd tick to avoid crowding
plt.grid(True)
plt.show()
```

##### 2.Line Chart of Job Satisfaction (`JobSatPoints_6`) Across Experience Levels

- Create a line chart to explore trends in job satisfaction (`JobSatPoints_6`) based on experience level.

- This chart will provide insight into how satisfaction correlates with experience over time


```python
## Write your code here
df_exp_jobsat = df[['YearsCodePro', 'JobSatPoints_6']].dropna()
median_jobsat_by_exp = df_exp_jobsat.groupby('YearsCodePro')['JobSatPoints_6'].median().reset_index()
plt.figure(figsize=(8,5))

plt.plot(
    median_jobsat_by_exp['YearsCodePro'], 
    median_jobsat_by_exp['JobSatPoints_6'], 
    marker='o', 
    linestyle='-', 
    color='teal'
)

plt.title("Median Job Satisfaction Across Experience Levels")
plt.xlabel("Years of Professional Experience")
plt.ylabel("Median Job Satisfaction")
plt.xticks(median_jobsat_by_exp['YearsCodePro'][::2], rotation=45)  # Show every 2nd tick to avoid crowding
plt.grid(True)
plt.show()
```
