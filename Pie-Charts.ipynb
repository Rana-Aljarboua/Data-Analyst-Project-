### Task 1: Visualizing Data Composition with Pie Charts


##### 1.1 Create a Pie Chart of the Top 5 Databases Respondents Want to Work With


In the survey data, the `DatabaseWantToWorkWith` column lists the databases that respondents wish to work with. Let’s visualize the top 5 most-desired databases in a pie chart.



```python
##Write your code here
from collections import Counter
db_data = df['DatabaseWantToWorkWith'].dropna()
db_list = db_data.str.split(';')
all_dbs = [db for sublist in db_list for db in sublist]
db_counter = Counter(all_dbs)

top5_dbs = db_counter.most_common(5)
print(top5_dbs)
labels, sizes = zip(*top5_dbs)

plt.figure(figsize=(8,8))
plt.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%', 
    startangle=140,       
    colors=['gold','lightcoral','lightskyblue','lightgreen','violet'] 
)

plt.title("Top 5 Databases Respondents Want to Work With")
plt.show()
```

The `DevType` column lists the developer types for respondents. We’ll examine the distribution by showing the top 5 developer roles in a pie chart.



```python
##Write your code here
dev_data = df['DevType'].dropna()
dev_list = dev_data.str.split(';')
all_roles = [role for sublist in dev_list for role in sublist]
role_counts = Counter(all_roles)
top5_roles = role_counts.most_common(5)
top5_roles
labels, sizes = zip(*top5_roles)

plt.figure(figsize=(8,8))
plt.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%',
    startangle=140
)

plt.title("Top 5 Developer Roles")
plt.show()
```

##### 1.3 Create a pie chart for the operating systems used by respondents for professional use


The `OpSysProfessional` use column shows the operating systems developers use professionally. Let’s visualize the distribution of the top operating systems in a pie chart.



```python
##Write your code here
opsys_data = df['OpSysProfessional use'].dropna()
opsys_list = opsys_data.str.split(';')
all_systems = [system for sublist in opsys_list for system in sublist]
opsys_counts = Counter(all_systems)
top_opsys = opsys_counts.most_common(5)
top_opsys
labels, sizes = zip(*top_opsys)

plt.figure(figsize=(8,8))
plt.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%',
    startangle=140
)

plt.title("Top Operating Systems Used Professionally")
plt.show()
```

##### 3.3 Pie Chart for Preferred Web Frameworks


The `WebframeWantToWorkWith` column includes web frameworks that respondents are interested in working with. Visualize the top 5 frameworks in a pie chart.



```python
##Write your code here
web_data = df['WebframeWantToWorkWith'].dropna()

# 2) نفصل القيم اللي مفصولة بـ ";"
web_list = web_data.str.split(';')

# 3) نفك القوائم (flatten) عشان يصير عندنا قائمة واحدة كبيرة
all_web = [fw for sublist in web_list for fw in sublist]

# 4) نحسب أكثر 5 أطر عمل
web_counts = Counter(all_web)
top5_web = web_counts.most_common(5)

# 5) نفصل الأسماء والأرقام عشان Pie Chart
labels, sizes = zip(*top5_web)

# 6) نرسم Pie Chart
plt.figure(figsize=(8,8))
plt.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%',   # يطلع النسبة لكل قطعة
    startangle=140
)

plt.title("Top 5 Web Frameworks Respondents Want to Work With")
plt.show()
```
