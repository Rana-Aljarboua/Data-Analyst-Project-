#### Step 1: Import the necessary module.


### 1. Load the Dataset


<h5>1.1 Import necessary libraries and load the dataset.</h5>


Ensure the dataset is loaded correctly by displaying the first few rows.


```python
# Import necessary libraries
import pandas as pd

# Load the Stack Overflow survey data
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(dataset_url)

# Display the first few rows
print(df.head())
```

#### 2. Explore the Dataset


<h5>2.1 Summarize the dataset by displaying the column data types, counts, and missing values.</h5>


```python
# Write your code here
print(df.info())
missing_values = df.isna().sum()
print(missing_values)

# summary
summary = pd.DataFrame({
    'dtype': df.dtypes,
    'Non-Null Count': df.count(),
    'Missing Values': missing_values
})

print(summary)
```

<h5>2.2 Generate basic statistics for numerical columns.</h5>


```python
# Write your code here
df.describe()
```

### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# Write your code here
df['Country'] = df['Country'].fillna('Unknown')

country_counts = df['Country'].value_counts()
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
## Write your code here
country_counts = df['EdLevel'].value_counts()
print(country_counts)
edlevel_map = {
    "Bachelor’s degree (B.A., B.S., B.Eng., etc.)": "Bachelor’s",
    "Master’s degree (M.A., M.S., M.Eng., MBA, etc.)": "Master’s",
    "Some college/university study without earning a degree": "Some College",
    "Secondary school (e.g. American high school, German Realschule or Gymnasium, etc.)": "Secondary",
    "Professional degree (JD, MD, Ph.D, Ed.D, etc.)": "Professional / Doctoral",
    "Associate degree (A.A., A.S., etc.)": "Associate",
    "Primary/elementary school": "Primary",
    "Something else": "Other"
}
df['EdLevel'] = df['EdLevel'].replace(edlevel_map)
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
## Write your code here
df['Employment_list'] = df['Employment'].str.split(';')
df_exploded = df.explode('Employment_list')
employment_dummies = pd.get_dummies(df_exploded['Employment_list'])
employment_encoded = employment_dummies.groupby(df_exploded.index).max()

df = pd.concat([df, employment_encoded], axis=1)
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here
missing_counts = df.isna().sum()
missing_sorted = missing_counts.sort_values(ascending=False)

print(missing_sorted.head(10)) 
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here
df['ConvertedCompYearly'].fillna(df['ConvertedCompYearly'].median())
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here
df['RemoteWork'].fillna(df['RemoteWork'].mode()[0])
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here
min_salary = df['ConvertedCompYearly'].min()
max_salary = df['ConvertedCompYearly'].max()

df['ConvertedCompYearly_MinMax'] = (
    (df['ConvertedCompYearly'] - min_salary) / 
    (max_salary - min_salary)
)
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
## Write your code here
import numpy as np

df['ConvertedCompYearly_Log'] = np.log1p(df['ConvertedCompYearly'])
df['ConvertedCompYearly'].skew()
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
## Write your code here
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')

df['YearsCodePro'].fillna(0, inplace=True)

bins = [0, 1, 4, 9, 19, float('inf')]
labels = ['Beginner', 'Junior', 'Mid-level', 'Senior', 'Expert']

df['ExperienceLevel'] = pd.cut(df['YearsCodePro'], bins=bins, labels=labels)

df[['YearsCodePro', 'ExperienceLevel']].head(10)
```
